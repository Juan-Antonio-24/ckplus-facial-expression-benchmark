```python
from pathlib import Path
import shutil
```

```python
dataset_dir = Path("../dataset/CK+")
images_dir = dataset_dir / "cohn-kanade-images"
labels_dir = dataset_dir / "Emotion"
coordinates_dir = dataset_dir / "Landmarks"
```

```python
prepared_dataset = Path("../dataset/CK+_prepared")
prepared_dataset.mkdir(parents=True, exist_ok=True)

coordinates_dataset = Path("../dataset/CK+_coordinates")
coordinates_dataset.mkdir(parents=True, exist_ok=True)
```

```python
def prepare_data(input_route, output_route, location_list, extension):
    excluded_files = {"S129_002_00000011", "S129_002_00000011_landmarks"}

    for subject_dir in input_route.iterdir():
        if not subject_dir.is_dir():
            continue

        for sequence_dir in subject_dir.iterdir():
            if not sequence_dir.is_dir():
                continue

            subject = subject_dir.name
            sequence = sequence_dir.name

            emotion_sequence_dir = labels_dir / subject / sequence

            if not emotion_sequence_dir.exists():
                continue

            txt_files = list(emotion_sequence_dir.glob("*.txt"))

            if not txt_files:
                continue

            with open(txt_files[0], "r") as file:
                label = int(float(file.read().strip()))

            files = sorted(sequence_dir.glob(f"*.{extension}"), key=lambda x: int(x.stem.split("_")[location_list]))

            if not files:
                continue

            first_file = files[0]
            last_file = files[-1]

            neutral_dir = output_route / "0"
            neutral_dir.mkdir(parents=True, exist_ok=True)

            neutral_name = f"{subject}_{sequence}_neutral{first_file.suffix}"

            shutil.copy2(first_file, neutral_dir / neutral_name)

            if last_file.stem in excluded_files:
                continue

            emotion_output_dir = output_route / str(label)
            emotion_output_dir.mkdir(parents=True, exist_ok=True)

            emotion_name = f"{subject}_{sequence}{last_file.suffix}"

            shutil.copy2(last_file, emotion_output_dir / emotion_name)

            print(f"{subject}/{sequence} -> neutral: 0 | emotion: {label}")

    print("Separation complete.")
```

```python
excluded_files = {"S129_002_00000011", "S129_002_00000011_emotion"}
```

```python
dataset_separate = Path("../dataset/CK+_prepared")
dataset_landmarks = Path("../dataset/CK+_coordinates")

def num_files(dataset_route, extension, data_type):
    labels = []
    for label in dataset_route.iterdir():
        if label.is_dir():
            labels.append(int(label.name))

    sorted_labels = sorted(labels)

    for number in sorted_labels:
        file = dataset_route / str(number)
        if file.is_dir():
            files = list(file.glob(f"*.{extension}")) 
            print(f"Emotion {file.name} has {len(files)} {data_type}.")
```

```python
prepare_data(images_dir, prepared_dataset, -1, "png")
```

```python
prepare_data(coordinates_dir, coordinates_dataset, -2, "txt")
```

```python
num_files(dataset_separate, "png", "images")
```

```python
num_files(coordinates_dataset, "txt", "text files")
```
